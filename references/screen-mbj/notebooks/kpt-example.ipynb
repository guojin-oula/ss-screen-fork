```python
import numpy as np
def find_kcoord(name, labels, kpoints):
    """
    Find the k coordinates from a given list of tuples containing (idx, name)
    """
    idx = None
    for i, sym in labels:
        if sym == name:
            idx = i
    if idx is None:
        raise ValueError(f"Cannot found point {name}")
    return idx, kpoints[idx]
    
def cut_ksegment(new_path, labels, kpts, nreduce=2):
    """
    Select a sepecific segment from and existing kpoint path
    :param new_path: A list of high symmetry point labels in the new path
    :param labels: Original labels of high symmetry kpoints as returnbed BandsData.labels
    :param kpts: Original kpoint coorindates
    :param nreduce: Factor to reduce the kpoint sampling along the path
    :returns: A tuple of new kpoints and new labels to be used for a KpointsData
    """
    d = np.linalg.norm(kpts[1] - kpts[0]) * 2
    new_labels = []
    new_points = []
    for i, sym in enumerate(new_path[:-1]):
        new_labels.append((len(new_points) , sym))
        start = find_kcoord(sym, labels, kpts)[1]
        next_sym = new_path[i+1]
        end = find_kcoord(next_sym, labels, kpts)[1]
        npts = int(np.ceil(np.linalg.norm(end-start) / d)) # Compute the number of points - keeping original separation
        new_ks = np.linspace(start, end, npts+1)
        new_points.extend(new_ks[:-1])
        if i == len(new_path) -2:
            new_labels.append((len(new_points) , new_path[-1]))
            new_points.append(end)
    new_points = np.stack(new_points, axis=0)
    return new_points, new_labels
```
