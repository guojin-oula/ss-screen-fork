```python
from ase import Atoms
from ase.optimize import LBFGSLineSearch, FIRE
from ase.filters import ExpCellFilter
from ase.constraints import Filter, FixSymmetry
from mace.calculators.foundations_models import mace_mp
```

```python
calc = mace_mp()
```

```python
prim = Atoms(
    'Ca3SnO',
    scaled_positions=[[0.5, 0.5, 0], [0, 0.5, 0.5], [0.5, 0, 0.5], [0, 0, 0], [0.5, 0.5, 0.5]], cell=[4.8,4.8,4.8],
pbc=True)
```

```python
subs = []
for Asite in ['Mg', 'Ca', 'Sr', 'Ba']:
    for Bsite in ['Si', 'Ge', 'Sb', 'Pb']:
        for Csite in ['O', 'S', 'Se', 'Te']:
            sub = prim.copy()
            sub.symbols[sub.symbols == 'Ca' ] = Asite
            sub.symbols[sub.symbols == 'Sn' ] = Bsite
            sub.symbols[sub.symbols == 'O' ] = Csite
            subs.append(sub)
```

```python
def relax_full(atoms):
    atoms = atoms.copy()
    atoms.calc = calc
    atoms.set_constraint(FixSymmetry(atoms))
    opt = FIRE(ExpCellFilter(atoms))
    opt.run(0.01)
    return atoms
```

```python
subs_relaxed = [relax_full(atoms) for atoms in subs]
```

```python
for atoms in subs_relaxed:
    print(atoms.symbols.get_chemical_formula(), atoms.get_volume())
```
